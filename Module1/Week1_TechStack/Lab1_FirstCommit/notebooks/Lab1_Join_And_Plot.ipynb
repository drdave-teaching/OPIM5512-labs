```python
import pandas as pd
import matplotlib.pyplot as plt

# Point this at YOUR pair repo, on the main branch.
# github.com URL -> click Raw -> copy that URL -> keep everything up to /data/clean
REPO_RAW = "https://raw.githubusercontent.com/<your-org>/<your-repo>/main/data/clean"

weather = pd.read_csv(f"{REPO_RAW}/weather_hourly.csv", parse_dates=["hour"])
demand  = pd.read_csv(f"{REPO_RAW}/demand_hourly.csv",  parse_dates=["hour"])

print("weather", weather.shape, "| demand", demand.shape)
```

## 1. Join them

One line - *if* you both got the timestamp convention right.

```python
df = demand.merge(weather, on="hour", how="inner")
print(len(df), "matched hours out of 744")
df.head()
```

## 2. Demand vs temperature

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(df["temp_f"], df["load_mw"], s=12, alpha=0.5)
ax.set_xlabel("Temperature at the airport (deg F)")
ax.set_ylabel("New England demand (MW)")
ax.set_title("Does the weather move the grid?")
plt.show()

print("correlation:", round(df["temp_f"].corr(df["load_mw"]), 3))
```

## 3. The grid's heartbeat

```python
by_hour = df.assign(h=df["hour"].dt.hour).groupby("h")["load_mw"].mean()

fig, ax = plt.subplots(figsize=(7, 4))
by_hour.plot(ax=ax, marker="o")
ax.set_xlabel("Hour of day")
ax.set_ylabel("Average demand (MW)")
ax.set_title("When does New England use electricity?")
plt.show()
```
