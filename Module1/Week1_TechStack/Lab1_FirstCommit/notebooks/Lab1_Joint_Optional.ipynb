This joins your two files on `hour` and makes the one plot your manager actually wants: **demand vs temperature**.
It saves `temp_vs_load.png` — drag it into `images/` and it drops straight into `REPORT.md`.


````python
import pandas as pd, numpy as np

CAMPUS = "hartford"   # or "stamford"
base = "https://raw.githubusercontent.com/drdave-teaching/opim5512-lab1-template/main/data/clean"
wx  = pd.read_csv(f"{base}/weather_hourly_{CAMPUS}.csv", parse_dates=["hour"])
dem = pd.read_csv(f"{base}/demand_hourly.csv", parse_dates=["hour"])

df = wx.merge(dem, on="hour", how="inner")     # neither half could do this alone
print(len(df), "matched hours")

r = df["temp_f"].corr(df["load_mw"])
slope, _ = np.polyfit(df["temp_f"], df["load_mw"], 1)
print(f"correlation r = {r:.2f}   |   each +1 F is worth about {slope:,.0f} MW")

ax = df.plot.scatter(x="temp_f", y="load_mw", alpha=0.4, figsize=(8, 6),
                     title=f"Demand climbs with temperature (r = {r:.2f})")
ax.set_xlabel("temperature (F)"); ax.set_ylabel("demand (MW)")
ax.get_figure().savefig("temp_vs_load.png", dpi=150, bbox_inches="tight")

from google.colab import files
files.download("temp_vs_load.png")
````
